```python
%%writefile qsim.cuh
#pragma once
#include <cuda_runtime.h>
#include <cuComplex.h>
#include <cstdio>
#include <cstdlib>

#define CUDA_CHECK(call)                                                       \
    do {                                                                       \
        cudaError_t err = call;                                                \
        if (err != cudaSuccess) {                                              \
            fprintf(stderr, "CUDA error at %s:%d: %s\n", __FILE__, __LINE__,   \
                    cudaGetErrorString(err));                                  \
            exit(EXIT_FAILURE);                                                \
        }                                                                      \
    } while (0)

struct Gate2x2 {
    cuFloatComplex u00, u01, u10, u11;
};

__global__ void apply_gate_1q(cuFloatComplex* state, int n_qubits, int target, Gate2x2 g);
__global__ void apply_cnot(cuFloatComplex* state, int n_qubits, int ctrl, int tgt);
```

```python
%%writefile kernels.cu
#include "qsim.cuh"

__global__ void apply_gate_1q(cuFloatComplex* state, int n_qubits, int target, Gate2x2 g) {
    unsigned long long tid = blockIdx.x * (unsigned long long)blockDim.x + threadIdx.x;
    unsigned long long n_pairs = 1ULL << (n_qubits - 1);
    if (tid >= n_pairs) return;

    // Insert a '0' bit at position `target` to get idx0, and '1' for idx1
    unsigned long long low_mask = (1ULL << target) - 1;
    unsigned long long idx0 = ((tid & ~low_mask) << 1) | (tid & low_mask);
    unsigned long long idx1 = idx0 | (1ULL << target);

    cuFloatComplex a0 = state[idx0];
    cuFloatComplex a1 = state[idx1];

    state[idx0] = cuCaddf(cuCmulf(g.u00, a0), cuCmulf(g.u01, a1));
    state[idx1] = cuCaddf(cuCmulf(g.u10, a0), cuCmulf(g.u11, a1));
}

__global__ void apply_cnot(cuFloatComplex* state, int n_qubits, int ctrl, int tgt) {
    unsigned long long tid = blockIdx.x * (unsigned long long)blockDim.x + threadIdx.x;
    unsigned long long n_pairs = 1ULL << (n_qubits - 1);
    if (tid >= n_pairs) return;

    unsigned long long low_mask = (1ULL << tgt) - 1;
    unsigned long long idx0 = ((tid & ~low_mask) << 1) | (tid & low_mask);
    unsigned long long idx1 = idx0 | (1ULL << tgt);

    // Flip target amplitudes only when the control bit is 1
    if ((idx0 >> ctrl) & 1ULL) {
        cuFloatComplex tmp = state[idx0];
        state[idx0] = state[idx1];
        state[idx1] = tmp;
    }
}
```

```python
%%writefile main.cu
#include "qsim.cuh"
#include <algorithm>
#include <chrono>
#include <cstdlib>
#include <vector>

void cpu_hadamard(std::vector<cuFloatComplex>& state, int n_qubits, int target) {
    const float inv_sqrt2 = 0.70710678118f;
    unsigned long long n_pairs = 1ULL << (n_qubits - 1);
    unsigned long long low_mask = (1ULL << target) - 1;

    for (unsigned long long tid = 0; tid < n_pairs; ++tid) {
        unsigned long long idx0 = ((tid & ~low_mask) << 1) | (tid & low_mask);
        unsigned long long idx1 = idx0 | (1ULL << target);
        cuFloatComplex a0 = state[idx0];
        cuFloatComplex a1 = state[idx1];
        state[idx0] = make_cuFloatComplex((a0.x + a1.x) * inv_sqrt2, (a0.y + a1.y) * inv_sqrt2);
        state[idx1] = make_cuFloatComplex((a0.x - a1.x) * inv_sqrt2, (a0.y - a1.y) * inv_sqrt2);
    }
}

int main(int argc, char** argv) {
    int n_qubits = (argc > 1) ? std::atoi(argv[1]) : 22;
    unsigned long long dim = 1ULL << n_qubits;
    size_t bytes = dim * sizeof(cuFloatComplex);

    std::vector<cuFloatComplex> h_state(dim, make_cuFloatComplex(0.0f, 0.0f));
    h_state[0] = make_cuFloatComplex(1.0f, 0.0f);

    cuFloatComplex* d_state = nullptr;
    CUDA_CHECK(cudaMalloc(&d_state, bytes));
    CUDA_CHECK(cudaMemcpy(d_state, h_state.data(), bytes, cudaMemcpyHostToDevice));

    const float inv_sqrt2 = 0.70710678118f;
    Gate2x2 H = {
        make_cuFloatComplex(inv_sqrt2, 0.0f), make_cuFloatComplex(inv_sqrt2, 0.0f),
        make_cuFloatComplex(inv_sqrt2, 0.0f), make_cuFloatComplex(-inv_sqrt2, 0.0f)
    };

    int threads = 256;
    int blocks = (int)(((dim >> 1) + threads - 1) / threads);

    // 1. Verify GHZ circuit on GPU
    apply_gate_1q<<<blocks, threads>>>(d_state, n_qubits, 0, H);
    for (int q = 0; q < n_qubits - 1; ++q) {
        apply_cnot<<<blocks, threads>>>(d_state, n_qubits, q, q + 1);
    }
    CUDA_CHECK(cudaMemcpy(h_state.data(), d_state, bytes, cudaMemcpyDeviceToHost));

    float p0 = h_state[0].x * h_state[0].x + h_state[0].y * h_state[0].y;
    float p_last = h_state[dim - 1].x * h_state[dim - 1].x + h_state[dim - 1].y * h_state[dim - 1].y;

    // 2. Benchmark Hadamard layer: GPU vs CPU
    std::fill(h_state.begin(), h_state.end(), make_cuFloatComplex(0.0f, 0.0f));
    h_state[0] = make_cuFloatComplex(1.0f, 0.0f);
    CUDA_CHECK(cudaMemcpy(d_state, h_state.data(), bytes, cudaMemcpyHostToDevice));

    cudaEvent_t start, stop;
    CUDA_CHECK(cudaEventCreate(&start));
    CUDA_CHECK(cudaEventCreate(&stop));
    CUDA_CHECK(cudaEventRecord(start));
    for (int q = 0; q < n_qubits; ++q) {
        apply_gate_1q<<<blocks, threads>>>(d_state, n_qubits, q, H);
    }
    CUDA_CHECK(cudaEventRecord(stop));
    CUDA_CHECK(cudaEventSynchronize(stop));
    float gpu_ms = 0.0f;
    CUDA_CHECK(cudaEventElapsedTime(&gpu_ms, start, stop));

    auto cpu_t0 = std::chrono::high_resolution_clock::now();
    for (int q = 0; q < n_qubits; ++q) {
        cpu_hadamard(h_state, n_qubits, q);
    }
    auto cpu_t1 = std::chrono::high_resolution_clock::now();
    double cpu_ms = std::chrono::duration<double, std::milli>(cpu_t1 - cpu_t0).count();

    printf("RESULT n=%d p0=%.4f plast=%.4f gpu_ms=%.3f cpu_ms=%.3f speedup=%.2f\n",
           n_qubits, p0, p_last, gpu_ms, cpu_ms, cpu_ms / gpu_ms);

    CUDA_CHECK(cudaEventDestroy(start));
    CUDA_CHECK(cudaEventDestroy(stop));
    CUDA_CHECK(cudaFree(d_state));
    return 0;
}
```

```python
import re
import subprocess
import matplotlib.pyplot as plt

qubits = list(range(16, 25))
gpu_times, cpu_times, speedups = [], [], []

for n in qubits:
    out = subprocess.check_output(["./qsim", str(n)], text=True)
    m = re.search(r"gpu_ms=([\d.]+) cpu_ms=([\d.]+) speedup=([\d.]+)", out)
    gpu_times.append(float(m.group(1)))
    cpu_times.append(float(m.group(2)))
    speedups.append(float(m.group(3)))
    print(f"{n} qubits -> GPU: {m.group(1)} ms | CPU: {m.group(2)} ms | Speedup: {m.group(3)}x")

plt.figure(figsize=(8, 4.5))
plt.plot(qubits, speedups, marker="o", linewidth=2)
plt.title("CuQSim: GPU vs. CPU Speedup by Qubit Count (T4 GPU)")
plt.xlabel("Number of Qubits")
plt.ylabel("Speedup (x)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("speedup.png", dpi=150)
plt.show()
```

```python
import cupy as cp
import numpy as np
from cuquantum import ComputeType, cudaDataType
from cuquantum.bindings import custatevec as cusv

def bench_custatevec(n_qubits: int) -> float:
    dim = 1 << n_qubits
    d_sv = cp.zeros(dim, dtype=cp.complex64)
    d_sv[0] = 1.0 + 0.0j

    inv_sqrt2 = np.float32(1.0 / np.sqrt(2.0))
    h_gate = np.array(
        [[inv_sqrt2, inv_sqrt2], [inv_sqrt2, -inv_sqrt2]],
        dtype=np.complex64,
    )

    handle = cusv.create()
    ws_size = cusv.apply_matrix_get_workspace_size(
        handle,
        cudaDataType.CUDA_C_32F,
        n_qubits,
        h_gate.ctypes.data,
        cudaDataType.CUDA_C_32F,
        cusv.MatrixLayout.ROW,
        0,  # adjoint = False
        1,  # n_targets = 1
        0,  # n_controls = 0
        ComputeType.COMPUTE_32F,
    )
    ws = cp.cuda.alloc(ws_size) if ws_size > 0 else None
    ws_ptr = ws.ptr if ws is not None else 0

    # Warmup
    cusv.apply_matrix(
        handle, d_sv.data.ptr, cudaDataType.CUDA_C_32F, n_qubits,
        h_gate.ctypes.data, cudaDataType.CUDA_C_32F, cusv.MatrixLayout.ROW,
        0, [0], 1, [], [], 0, ComputeType.COMPUTE_32F, ws_ptr, ws_size,
    )
    cp.cuda.Device().synchronize()

    # Timed Hadamard layer
    start, stop = cp.cuda.Event(), cp.cuda.Event()
    start.record()
    for q in range(n_qubits):
        cusv.apply_matrix(
            handle, d_sv.data.ptr, cudaDataType.CUDA_C_32F, n_qubits,
            h_gate.ctypes.data, cudaDataType.CUDA_C_32F, cusv.MatrixLayout.ROW,
            0, [q], 1, [], [], 0, ComputeType.COMPUTE_32F, ws_ptr, ws_size,
        )
    stop.record()
    stop.synchronize()

    ms = cp.cuda.get_elapsed_time(start, stop)
    cusv.destroy(handle)
    return float(ms)

print(f"24-Qubit Hadamard Layer (cuStateVec): {bench_custatevec(24):.3f} ms")
```

```python
import re
import subprocess
import matplotlib.pyplot as plt

qubits = list(range(16, 25))
my_gpu_ms, cusv_ms = [], []

for n in qubits:
    out = subprocess.check_output(["./qsim", str(n)], text=True)
    m = re.search(r"gpu_ms=([\d.]+)", out)
    t_my = float(m.group(1))
    t_nv = bench_custatevec(n)
    my_gpu_ms.append(t_my)
    cusv_ms.append(t_nv)
    print(f"{n} qubits -> CuQSim: {t_my:.3f} ms | cuStateVec: {t_nv:.3f} ms | Ratio: {t_my / t_nv:.2f}x")

plt.figure(figsize=(8, 4.5))
plt.plot(qubits, my_gpu_ms, marker="o", label="CuQSim (Custom CUDA)")
plt.plot(qubits, cusv_ms, marker="s", label="NVIDIA cuStateVec")
plt.yscale("log")
plt.title("Hadamard Layer Runtime: Custom CUDA Kernel vs. cuStateVec (T4 GPU)")
plt.xlabel("Number of Qubits")
plt.ylabel("Runtime (ms, log scale)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("cuqsim_vs_custatevec.png", dpi=150)
plt.show()
```
